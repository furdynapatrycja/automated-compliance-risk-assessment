
# Automated Compliance Risk Assessment & Register

## 1. Project Overview

This project demonstrates a structured Governance, Risk and Compliance (GRC) approach to identifying, assessing, monitoring and prioritizing compliance and operational risks.

The solution uses Python and Pandas to build a risk register, map business processes to regulatory requirements, assess control effectiveness, calculate residual risk and generate remediation priorities.

## 2. Fictional Company Profile

**Company Name:** FinoraPay  
**Industry:** Financial Technology (FinTech)  
**Headquarters:** Warsaw, Poland  
**Employees:** 250  
**Customers:** 100,000  
**Business Model:** Digital payment services for individual and business customers  
**Technology Environment:** Cloud-based infrastructure

## 3. Project Objectives

- Identify key business processes and associated risks.
- Establish a regulatory and standards universe.
- Map risks to business processes and controls.
- Calculate inherent and residual risk scores.
- Assess control effectiveness.
- Prioritize remediation activities.
- Generate management-level risk reporting.

## 4. Scope and Limitations

FinoraPay is a fictional organization created for educational and portfolio purposes.

Risk scores and control effectiveness values are illustrative assumptions. They do not represent a real audit, legal opinion, regulatory assessment or certification.

Regulatory applicability must be assessed based on the actual legal status, activities and circumstances of an organization.

```python

risk_assessment = risk_register.merge(
    control_inventory[
        ["control_id", "control_name", "effectiveness"]
    ],
    on="control_id",
    how="left"
)

display(risk_assessment)
```

```python

risk_assessment["inherent_risk_score"] = (
    risk_assessment["likelihood"]
    * risk_assessment["impact"]
    * risk_assessment["velocity"]
)

risk_assessment["residual_risk_score"] = (
    risk_assessment["inherent_risk_score"]
    * (1 - risk_assessment["effectiveness"])
)

risk_assessment["residual_risk_score"] = (
    risk_assessment["residual_risk_score"].round(2)
)

display(
    risk_assessment[
        [
            "risk_id",
            "risk_name",
            "inherent_risk_score",
            "effectiveness",
            "residual_risk_score"
        ]
    ]
)
```

```python

def classify_risk(score):
    if score >= 80:
        return "Critical"
    elif score >= 50:
        return "High"
    elif score >= 25:
        return "Medium"
    else:
        return "Low"


risk_assessment["inherent_risk_level"] = (
    risk_assessment["inherent_risk_score"].apply(classify_risk)
)

risk_assessment["residual_risk_level"] = (
    risk_assessment["residual_risk_score"].apply(classify_risk)
)

display(
    risk_assessment[
        [
            "risk_id",
            "risk_name",
            "inherent_risk_level",
            "residual_risk_level"
        ]
    ]
)
```

```python

total_risks = len(risk_assessment)

high_critical_risks = risk_assessment[
    risk_assessment["residual_risk_level"].isin(
        ["High", "Critical"]
    )
].shape[0]

average_inherent_risk = (
    risk_assessment["inherent_risk_score"].mean()
)

average_residual_risk = (
    risk_assessment["residual_risk_score"].mean()
)

open_actions = remediation_tracker[
    remediation_tracker["status"] != "Closed"
].shape[0]

management_summary = pd.DataFrame([
    {"metric": "Total Risks", "value": total_risks},
    {"metric": "High and Critical Residual Risks", "value": high_critical_risks},
    {"metric": "Average Inherent Risk", "value": round(average_inherent_risk, 2)},
    {"metric": "Average Residual Risk", "value": round(average_residual_risk, 2)},
    {"metric": "Open Remediation Actions", "value": open_actions}
])

display(management_summary)
```

```python

risk_counts = risk_assessment[
    "residual_risk_level"
].value_counts().reindex(
    ["Low", "Medium", "High", "Critical"],
    fill_value=0
)

plt.figure(figsize=(9, 5))

risk_counts.plot(
    kind="bar",
    title="Residual Risk Distribution"
)

plt.xlabel("Residual Risk Level")
plt.ylabel("Number of Risks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

heatmap_data = risk_assessment.pivot_table(
    index="impact",
    columns="likelihood",
    values="risk_id",
    aggfunc="count",
    fill_value=0
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Inherent Risk Heatmap")
plt.xlabel("Likelihood")
plt.ylabel("Impact")

plt.tight_layout()

plt.savefig(
    "risk_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
